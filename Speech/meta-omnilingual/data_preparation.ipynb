```python
from datasets import load_dataset
```

```python
# Load the English Wikipedia dataset
wikipedia_dataset = load_dataset("wikimedia/wikipedia", "20231101.en")
wikipedia_dataset
```

```python
# Filter for articles where the phrase "machine learning" appears at least 3 times
def has_machine_learning_3times(example):
    text_lower = example["text"].lower()
    count = text_lower.count("machine learning")
    return count >= 3


machine_learning_articles = wikipedia_dataset["train"].filter(
    has_machine_learning_3times,
    desc="Filtering for 'machine learning' >= 3",
    batch_size=1000,
    writer_batch_size=1000,
    num_proc=4,  # tqdm only tracks one process
)
```

```python
machine_learning_articles
```

```python
import huggingface_hub

huggingface_hub.login()
```

```python
machine_learning_articles.push_to_hub("username/wiki-ai-filtered")
```
